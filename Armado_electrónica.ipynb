## Configuración de Autopiloto, Arduplane

El pixhawk 6c venía con el firmware de arducopter instalado, por lo que hay que cambiarlo a arduplane, ya que Quadplane (las configuraciones q queremos del vtol) son una opción de configuración de arduplane. Para eso es neccesario tener el software Mission Planer o PX4 instalado. En este caso se decidió utilizar Mission Planner dado su foco en usuario y no en desarrolladores como lo es PX4 (por ejemplo si se quisiera trabajar junto con otro controlador como Raspberry Pi se utilizaría PX4).
Para instalar el firmware es necesario conectar el dispositivo a través de cable USB-C al computador y verificar conección a algún puerto COM, presionar en el que se observe esta conección (por ejemplo si al desconectar ese puerto ya no está habilitado y al conectar nuevamente aparece), y configurar baud rate a 115200, previo a presionar "connect" hay que dirigirse la sección de setup, donde se presiona arduplane aceptar la instalación de firmware. Posteriormente se abre el cuadro de selección de "board" donde se selecciona "Pixhawk 6c", posteriormente hay que seguir con las instrucciones y esperar que el programa mencione "firmware updated", donde yo desconecté y volví a conectar la placa identificando que en la sección del puerto se obeserva que los que se activan no son los mismos vistos previamente, y se cambia a uno nuevo (en mi caso COM-13) donde al presionar "connect" se identifica que el firmware correcto esta instalado y la placa se comporta como es esperado.
Se utilizó la información encontrada en: https://ardupilot.org/plane/docs/common-loading-firmware-onto-pixhawk.html#common-loading-firmware-onto-pixhawk
