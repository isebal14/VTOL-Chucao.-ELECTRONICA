# Se utiliza la firmware de arduplane abilitando Quadcopter

# 1 Configuración de autopiloto
Dado que no se tenía conocimiento de las conecciones el las salidas PWM del autopiloto se busca la informadción del firmware a través de utilizar un Pixhawk 2 Cube del laboratorio.  
- Hay que abilitar la función de Q_ENABLE en la lista de los parámetros, set to 1.
- Los motores en los PWM output para el quadplane son desde el 5 en adelante.  
  
  
https://ardupilot.org/plane/docs/quadplane-parameters.html  
Ese link tiene la información acerca configuración de quadplane en firmware de fixed wing airplane.  


# 2 Información de motores
En la energía puede ser independiente enrte los quedplpane lift motors, y el pusher.  
Para el monitoreo de la batería se conecta el power module PM02 entre el motor pusher y batería, con el que se alimenta el autopiloto en power 1, y se requiere conectar en power 2 los pines de monitoreo de corriente en el autopiloto, previamente habilitando este puerto en la configuración según https://ardupilot.org/copter/docs/common-power-module-configuration-in-mission-planner.html  
Es requerido hacer calibración de esc con autopiloto, según https://ardupilot.org/plane/docs/quadplane-esc-calibration.html  



# 3 Configuración de servo outputs para sitemas de control  
Es importante determinar el output de los servos a los calanes correspondientes:  
- Roll: Output 1, en este se utiliza un servo splitter para que la señal sea invertida entre estos. Asegurar en la instalación el correcto funcionamiento.
- Yaw y Pitch: Se requiere configurar las salidas de los canales 2 y 4 como vtail left (2) y vtail right (4) para los cuales revisar movimiento deseado con radio y al seleccionar modo FBWA.

# 4 Proceso de enlace de receptor BR1 en caso 2
Para "bind" el receptor a la radio se siguen los pasos encontrados en https://manuals.plus/radiomaster/br1-receiver-manual.  
- 1ro: Mantener apagado TX
- 2do: Entregar y quitar energía del receptor 3 veces, y verificar doble blink de LED
- 3ro: Encender transmisor
- 4to: Seleccionar enlace en modo ExpressLRS
- 5to: Evaluar luz sólida de encendido
